```python
df['total_score']=df['math_score']+df['science_score']+df['programming_score']+df['english_score']

top_student=df.loc[df['total_score'].idxmax()]

print("Top student : ")
print(f"Name : {top_student['name']}")
print(f"Department : {top_student['department']}")
print(f"Total Score : {top_student['total_score']}")
print(f"Attendance : {top_student['attendance_percentage']}")
```

```python
dept_avg=df.groupby('department')['total_score'].mean().round(2)
print("Avg total score by department:")
print(dept_avg)
```

```python
df_sorted=df.sort_values(by='total_score',ascending=False)
print(df_sorted.head(10))
```

```python
#who is the student with the lowest programming score?
lowest=df[df['programming_score']==df['programming_score'].min()]
print(lowest['name'])
```

```python
#how many students score above 80 in math?
maths=df[df['math_score']>80]
print(len(maths))
```

```python
#What is the average total score for students with attendance above 90 percentage?
above90=df[df['attendance_percentage']>90]
avg_above90=above90['total_score'].mean()
print(avg_above90)
```
